```python
import pandas as pd
import numpy as np
```

```python
inferences = pd.read_csv("inferences.csv")
inferences = inferences.drop(columns=['Unnamed: 0'])
inferences.head()
```

```python
accuracy_gender = np.mean(inferences["gender"] == inferences["pred_gender"])  * 100
accuracy_age = np.mean(inferences["age"] == inferences["pred_age"]) * 100
accuracy_expression = np.mean(inferences["expression"] == inferences["pred_expression"]) * 100
print(accuracy_gender, accuracy_age, accuracy_expression)
```

```python
pd.DataFrame(columns=["Age", "Gender", "Expression"], data=[[accuracy_gender, accuracy_age, accuracy_expression]])
```
